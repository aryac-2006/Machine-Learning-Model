```python
# Q1. Data Preparation

import pandas as pd
from sklearn.model_selection import train_test_split

# Load dataset
df = pd.read_csv(
   "/content/train.txt",
    sep=";",
    names=["text", "emotion"]
)

# Remove missing values
df = df.dropna()

X = df["text"]
y = df["emotion"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# Q2. Bag of Words – Basic

from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)

print("Training matrix:", X_train_bow.shape)
print("\nTesting matrix :", X_test_bow.shape)

print("\n<><><><><><><><><><><><><> First 20 feature names <><><><><><><><><><><><>\n")
print(vectorizer.get_feature_names_out()[:20])
```

```python
# Q3. Bag of Words + MultinomialNB

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

nb_bow = MultinomialNB()

nb_bow.fit(X_train_bow, y_train)

y_pred_bow = nb_bow.predict(X_test_bow)

bow_accuracy = accuracy_score(y_test, y_pred_bow)

print("<<<< ---------- Bag of Words Accuracy ------------- >>>>\n")
print(bow_accuracy)
```

```python
# Q4. Understanding Vocabulary

vocabulary = vectorizer.get_feature_names_out()

print("<<< Total vocabulary size >>> :", len(vocabulary))

print("\n_________________________15 words from vocabulary_______________________\n")
print(vocabulary[:15])

# Convert one training document into BoW vector
sample_text = X_train.iloc[0]

sample_vector = vectorizer.transform([sample_text])

print("\n<<<< ---------------- Sample document ----------------- >>>>\n")
print(sample_text)

print("\n<<<< ------------------- BoW vector -------------------- >>>>\n")
print(sample_vector.toarray())
```

```python
# Q5. N-grams with Bag of Words

bigram_vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

X_train_bigram = bigram_vectorizer.fit_transform(X_train)
X_test_bigram = bigram_vectorizer.transform(X_test)

print("<<<< Bigram training matrix >>>> :", X_train_bigram.shape)
print("<<<< Bigram testing matrix >>>> :", X_test_bigram.shape)

print("\n____________________Some unigram + bigram features___________________\n")
print(bigram_vectorizer.get_feature_names_out()[:30])
```

```python
# Q6. N-grams + Model Training

nb_bigram = MultinomialNB()

nb_bigram.fit(X_train_bigram, y_train)

y_pred_bigram = nb_bigram.predict(X_test_bigram)

bigram_accuracy = accuracy_score(
    y_test,
    y_pred_bigram
)

print("<<< Unigram BoW Accuracy >>> :", bow_accuracy)
print("<<< Bigram BoW Accuracy  >>> :", bigram_accuracy)

if bigram_accuracy > bow_accuracy:
    print("\nBigram model performed better.")
elif bigram_accuracy < bow_accuracy:
    print("\nUnigram model performed better.")
else:
    print("\nBoth models performed equally.")
```

```python
# Q7. TF-IDF Vectorization

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("<<<< TF-IDF training matrix >>>> :", X_train_tfidf.shape)
print("\n<<<< TF-IDF testing matrix >>>> :", X_test_tfidf.shape)

print("\n_____________________{ First 15 TF-IDF features }_____________________\n")
print(tfidf_vectorizer.get_feature_names_out()[:15])
```

```python
# Q8. TF-IDF + MultinomialNB

nb_tfidf = MultinomialNB()

nb_tfidf.fit(X_train_tfidf, y_train)

y_pred_tfidf = nb_tfidf.predict(X_test_tfidf)

tfidf_accuracy = accuracy_score(
    y_test,
    y_pred_tfidf
)

print("<<<<<< TF-IDF Accuracy >>>>>> :", tfidf_accuracy)
```

```python
# Q9. Comparison of Vectorizers

print("<<<<<<------------- Vectorizer Comparison ------------->>>>>>>\n")

comparison = pd.DataFrame({
    "Vectorizer": [
        "Bag of Words (Unigrams)",
        "Bag of Words (Unigrams + Bigrams)",
        "TF-IDF"
    ],
    "Accuracy": [
        bow_accuracy,
        bigram_accuracy,
        tfidf_accuracy
    ]
})

print(comparison)

# Find the best method

best_index = comparison["Accuracy"].idxmax()

best_method = comparison.loc[
    best_index,
    "Vectorizer"
]

best_accuracy = comparison.loc[
    best_index,
    "Accuracy"
]

print("\n<<<< Best Method >>>> :", best_method)
print("<<<< Best Accuracy >>>> :", best_accuracy)
```

```python
# Q10. Mini Project – Complete Vectorization Pipeline

import joblib

# Compare all three approaches
accuracies = {
    "BoW": bow_accuracy,
    "Bigram BoW": bigram_accuracy,
    "TF-IDF": tfidf_accuracy
}

best_method = max(
    accuracies,
    key=accuracies.get
)

print("<<<<<<< ------------- Model Accuracies -------------- >>>>>>>\n")
for method, accuracy in accuracies.items():
    print(method, ":", accuracy)

print("\nBest method:", best_method)
print("Best accuracy:", accuracies[best_method])

# Save best vectorizer and model
if best_method == "BoW":
    best_vectorizer = vectorizer
    best_model = nb_bow

elif best_method == "Bigram BoW":
    best_vectorizer = bigram_vectorizer
    best_model = nb_bigram

else:
    best_vectorizer = tfidf_vectorizer
    best_model = nb_tfidf

joblib.dump(best_vectorizer, "best_vectorizer.pkl")
joblib.dump(best_model, "best_emotion_model.pkl")

print("\n<<<< Best vectorizer saved as >>>> : best_vectorizer.pkl")
print("<<<< Best model saved as >>>>      : best_emotion_model.pkl")
```
